```python
import sklearn
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split  # 알아서 학습
```

```python
import pandas as pd

iris = load_iris()

iris_data = iris.data # 피처 : 꽃바침의 너비 길이, 꽃잎의 너비 길이
iris_label = iris.target # 아이리스 품종 : 정답

print('iris target값 :', iris_label)
print('iris target명 :', iris.target_names)
```

```python
iris_df = pd.DataFrame(data= iris_data, columns= iris.feature_names)
iris_df['label'] = iris.target      # 레이블/타깃(label) : 맞히려는 정답 보통 y로 표기
iris_df.head()
```

```python
# 학습셋과 테스트셋으로 나누기
# train 학습 test 
X_train, X_test, Y_train, Y_test = train_test_split(iris_data, iris_label, test_size=0.2, random_state=39) # random_state 고정 시 매번 방식 같음
# test 크기 0.2 -> train 크기 0.8
```

```python
dt_clf = DecisionTreeClassifier(random_state=39) # 분류기
```

```python
dt_clf.fit(X_train, Y_train) # fit() - 학습시키기
```

```python
pred = dt_clf.predict(X_test) # predict() - 예측
pred
```

```python
from sklearn.metrics import accuracy_score
print('예측 정확도 :', accuracy_score(Y_test, pred)) # 정확도 1 일시 잘못된 코드
```

```python
from sklearn.model_selection import KFold # model_selection - 데이터 분리, 교차 검증 시 사용
import numpy as np                        # K-Fold란? 데이터를 k 조각 나눠 돌아가며 시험 - 교차 검증

kfold = KFold(n_splits= 5)
cv_accuracy = []
```

```python
n_iter = 0
features = iris.data
label = iris.target

for train_index, test_index in kfold.split(features) : # kfold로 나눈만큼 반복
    X_train, X_test = features[train_index], features[test_index]
    Y_train, Y_test = label[train_index], label[test_index]

    dt_clf.fit(X_train, Y_train)
    pred = dt_clf.predict(X_test)
    n_iter += 1

    accuracy = np.round(accuracy_score(Y_test, pred), 4)
    train_size = X_train.shape[0]
    test_size = X_test.shape[0]

    print(n_iter)
    print('교차검증 정확도 :', accuracy)
    print('학습 데이터 크기 :', train_size)
    print('검증 데이터 크기 :', test_size)
    cv_accuracy.append(accuracy)

print('평균 검증 정확도 : ', np.mean(cv_accuracy))
```

```python
kfold = KFold(n_splits= 3)

for train_index, test_index in kfold.split(iris_df):
    
    n_iter += 1
    
    label_train = iris_df['label'].iloc[train_index]
    label_test = iris_df['label'].iloc[test_index]

    print('학습 레이블 데이터 분포 :', label_train.value_counts())
    print('검증 레이블 데이터 분포 :', label_test.value_counts())
```

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits = 3)
n_iter = 0

for train_index, test_index in skf.split(iris_df, iris_df['label']) :
    n_iter += 1

    label_train = iris_df['label'].iloc[train_index] # 학습용 레이블
    label_test = iris_df['label'].iloc[test_index] # 검증용 레이블

    print('학습 레이블 데이터 분포 :', label_train.value_counts())
    print('검증 레이블 데이터 분포 :', label_test.value_counts())
```

```python
dt_clf = DecisionTreeClassifier(random_state= 156)
skfold = StratifiedKFold(n_splits=3)
n_iter = 0
cv_accuracy = [] # 각 fold별 정확도 저장할 리스트

for train_index, test_index in skfold.split(features, label):
    X_train, X_test = features[train_index], features[test_index] # 학습/검증용 분리
    Y_train, Y_test = label[train_index], label[test_index] # 학습/검증용 분리

    dt_clf.fit(X_train, Y_train)
    pred = dt_clf.predict(X_test) 

    n_iter += 1
    accuracy = np.round(accuracy_score(Y_test, pred), 4) # 정확도 소수점 4번째까지 출력
    cv_accuracy.append(accuracy)

print('평균 검증 정확도 :', np.round(np.mean(cv_accuracy), 4))
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(dt_clf, features, label, scoring='accuracy', cv = 3)

print('교차 검증 정확도 :', np.round(scores, 4))
print('평균 검증 정확도 :', np.round(np.mean(scores), 4))
```

```python
# gridsearchcv 하이퍼파라미터 최적화 도구

from sklearn.model_selection import GridSearchCV

X_train, X_test, Y_train, Y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state= 121)

parameters = {'max_depth' : [1, 2, 3], 'min_samples_split': [2, 3]}
```

```python
dtree = DecisionTreeClassifier()

grid_dtree = GridSearchCV(dtree, param_grid= parameters, cv=3, refit=True)

grid_dtree.fit(X_train, Y_train)
scores_df = pd.DataFrame(grid_dtree.cv_results_)
scores_df[['params', 'mean_test_score', 'rank_test_score', 'split0_test_score', 'split1_test_score', 'split2_test_score']]
```

```python
grid_dtree.best_params_ # 최고의 방법?
```

```python
grid_dtree.best_score_ # 위 최고의 방법으로 찾을 시 점수
```

```python
estimator = grid_dtree.best_estimator_

pred = estimator.predict(X_test)
print(accuracy_score(Y_test, pred))
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
iris_data = iris.data
iris_df = pd.DataFrame(data=iris_data, columns= iris.feature_names)

print(iris_df.mean()) # 평균
print(iris_df.var()) # 분산
```

```python
# StandardScaler

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)

iris_df_scaled = pd.DataFrame(data = iris_scaled, columns = iris.feature_names)

print(iris_df_scaled.mean())
print(iris_df_scaled.var())
```

```python
# MinMaxScaler

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)

iris_df_scaled = pd.DataFrame(data= iris_scaled, columns= iris.feature_names)
print(iris_df_scaled.min())
print(iris_df_scaled.max())
```
